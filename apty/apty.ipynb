```python
# Load the dataset
data_apty_ranked = load_dataset("worta/apty", "APTY-ranked")

# Convert to DataFrames
df_apty_ranked = pd.DataFrame(data_apty_ranked['train'])

print(df_apty_ranked.info())
```

```python
# Normalize the 'meta' column to create separate columns for 'id', 'annotators', and 'APT'
meta_df = pd.json_normalize(df_apty_ranked['meta'])

# Drop the original 'meta' column and concatenate the new columns
df_apty_ranked = df_apty_ranked.drop(columns=['meta'])
df_apty_ranked = pd.concat([df_apty_ranked, meta_df], axis=1)

# Display the updated DataFrame
print(df_apty_ranked.head())
```

```python
# Check the distribution of paraphrase types for the base dataset
print(df_apty_ranked['APT'].value_counts())
```

### Cleaning

```python
# Extract the text from the nested dictionaries for 'chosen' and 'rejected'
df_apty_ranked['original'] = df_apty_ranked['original'].apply(str)
df_apty_ranked['chosen'] = df_apty_ranked['chosen'].apply(lambda x: x['text'] if isinstance(x, dict) else str(x))
df_apty_ranked['rejected'] = df_apty_ranked['rejected'].apply(lambda x: x['text'] if isinstance(x, dict) else str(x))
print(df_apty_ranked.info())
```

```python
# Check for missing values
missing_values = df_apty_ranked.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Check for Leading or Trailing Whitespace
# Strip whitespace and compare to detect issues
df_apty_ranked['original'] = df_apty_ranked['original'].str.strip()
df_apty_ranked['chosen'] = df_apty_ranked['chosen'].str.strip()
df_apty_ranked['rejected'] = df_apty_ranked['rejected'].str.strip()

whitespace_issues = df_apty_ranked[['original', 'chosen', 'rejected']].apply(lambda x: x.str.contains(r'^\s+|\s+$')).any()
print("\nWhitespace Issues (True indicates issues):")
print(whitespace_issues)
```

```python
# Check for Inconsistent Punctuation (e.g., sentences not ending with a period)
punctuation_issues = df_apty_ranked[['original', 'chosen', 'rejected']].apply(lambda x: ~x.str.endswith('.'))
print("\nPunctuation Issues (True indicates issues):")
print(punctuation_issues.any())
```

```python
# Extract the last character of each sentence
df_apty_ranked['original_end'] = df_apty_ranked['original'].str[-1]
df_apty_ranked['chosen_end'] = df_apty_ranked['chosen'].str[-1]
df_apty_ranked['rejected_end'] = df_apty_ranked['rejected'].str[-1]

# Display sentences that do not end with a dot and show their ending character
non_ending_dot_original = df_apty_ranked[df_apty_ranked['original_end'] != '.']
non_ending_dot_chosen = df_apty_ranked[df_apty_ranked['chosen_end'] != '.']
non_ending_dot_rejected = df_apty_ranked[df_apty_ranked['rejected_end'] != '.']

print("Sentences in 'original' that do not end with a dot and their ending character:")
print(non_ending_dot_original[['original', 'original_end']].head())

print("\nSentences in 'chosen' that do not end with a dot and their ending character:")
print(non_ending_dot_chosen[['chosen', 'chosen_end']].head())

print("\nSentences in 'rejected' that do not end with a dot and their ending character:")
print(non_ending_dot_rejected[['rejected', 'rejected_end']].head())

# Display the unique ending characters that are not dots
unique_ending_characters_original = non_ending_dot_original['original_end'].unique()
unique_ending_characters_chosen = non_ending_dot_chosen['chosen_end'].unique()
unique_ending_characters_rejected = non_ending_dot_rejected['rejected_end'].unique()

print("\nUnique ending characters in 'original' that are not a dot:")
print(unique_ending_characters_original)

print("\nUnique ending characters in 'chosen' that are not a dot:")
print(unique_ending_characters_chosen)

print("\nUnique ending characters in 'rejected' that are not a dot:")
print(unique_ending_characters_rejected)

# Display count of such sentences
print("\nCount of sentences that do not end with a dot:")
print(f"Original: {len(non_ending_dot_original)}")
print(f"Chosen: {len(non_ending_dot_chosen)}")
print(f"Rejected: {len(non_ending_dot_rejected)}")
```

```python
# Function to modify the last character according to the rules
def modify_last_character(text):
    if text.endswith('"'):
        # Remove the last double quote
        text = text[:-1]
    elif text[-1].isalpha():
        # Add a '.' if the last character is a letter
        text += '.'
    return text
```

```python
# Apply the function to each column
df_apty_ranked['original'] = df_apty_ranked['original'].apply(modify_last_character)
df_apty_ranked['chosen'] = df_apty_ranked['chosen'].apply(modify_last_character)
df_apty_ranked['rejected'] = df_apty_ranked['rejected'].apply(modify_last_character)
```

```python
# Extract the last character of each sentence
df_apty_ranked['original_end'] = df_apty_ranked['original'].str[-1]
df_apty_ranked['chosen_end'] = df_apty_ranked['chosen'].str[-1]
df_apty_ranked['rejected_end'] = df_apty_ranked['rejected'].str[-1]

# Display sentences that do not end with a dot and show their ending character
non_ending_dot_original = df_apty_ranked[df_apty_ranked['original_end'] != '.']
non_ending_dot_chosen = df_apty_ranked[df_apty_ranked['chosen_end'] != '.']
non_ending_dot_rejected = df_apty_ranked[df_apty_ranked['rejected_end'] != '.']

# Display count of such sentences
print("\nCount of sentences that do not end with a dot:")
print(f"Original: {len(non_ending_dot_original)}")
print(f"Chosen: {len(non_ending_dot_chosen)}")
print(f"Rejected: {len(non_ending_dot_rejected)}")
```

```python
# Check for Special Characters (ensure there's no unwanted characters)
special_characters_issues = df_apty_ranked[['original', 'chosen', 'rejected']].apply(lambda x: x.str.contains(r'[ß]', regex=True).any())
print("\nSpecial Characters Issues (True indicates issues):")
print(special_characters_issues)
```

```python
# Check for Length Variation (identify any unusual length variations)
lengths = df_apty_ranked[['original', 'chosen', 'rejected']].applymap(len)
length_variation = lengths.std()
print("\nLength Variation (Standard Deviation) in Each Column:")
print(length_variation)
```

### Cleaning duplicates

```python
# Identify duplicate rows based on all columns
duplicate_rows = df_apty_ranked[df_apty_base_train.duplicated()]

# Count the number of duplicate rows
duplicate_count = len(duplicate_rows)
print(f"Total number of duplicate rows: {duplicate_count}")

# If you want to check for duplicates based on specific columns, you can specify them:
# For example, checking duplicates based on 'original' and 'APT' columns
duplicate_specific = df_apty_ranked[df_apty_ranked.duplicated(subset=['original', 'chosen', "rejected"])]

# Display only the 'original' and 'paraphrase_fixed' columns of these specific duplicate rows
duplicate_specific_columns = duplicate_specific[['original', 'chosen', "rejected"]]

# Count the number of duplicate rows based on specific columns
duplicate_specific_count = len(duplicate_specific_columns)
print(f"Total number of duplicate rows based on specific columns: {duplicate_specific_count}")
```

```python
df_unique_specific_columns = df_apty_ranked.drop_duplicates(subset=['original', 'chosen', "rejected"])

# Count the number of rows after dropping duplicates
unique_count = len(df_unique_specific_columns)
print(f"Total number of unique rows (based on specific columns): {unique_count}")

# Check the distribution of paraphrase types for the base dataset
print(df_unique_specific_columns['APT'].value_counts())
```

```python
# Rename the column 'original' to 'prompt'
df_unique_specific_columns = df_unique_specific_columns.rename(columns={'original': 'prompt'})

# Display the first few rows to confirm the change
print(df_unique_specific_columns.head())
```
